## What Is an Expression?

In this case, the string `"Hello, World!"` is an expression that evaluates to itself, a piece of text.

### Order of Operators

Python evaluates expressions in a fixed order (PEMDAS):

Parentheses: ()
Exponentiation: **
Multiplication, Division, Floor Division, Modulus: *, /, //, %
Addition and Subtraction: +, -

Operators on the same level are evaluated left to right.

```python
print(1 + 2 * 3)                     # 7
print((1 + 2) * 3)                   # 9
print(123456789 * 456789)            # 56393703190521
print(10 + 2 * 3 - 4 / 2)            # 14.0
print((20 % 3) + (5 ** 2 // 4))      # 8
print(((3 + 5) * 2 - 4) ** 2 / 8)    # 18.0
```

### Using Parentheses to Control Evaluation Order

Parentheses override Python’s default precedence rules. By grouping parts of an expression, you control the order in which calculations are performed.

Important notes

Left-to-right evaluation: operators with the same precedence are evaluated from left to right.
Exponentiation is right-associative: ** is evaluated from right to left.

By mastering operator precedence and using parentheses effectively, you can write clear and precise Python expressions that behave as intended.

```python
# Example 1: Addition, multiplication, and exponentiation
print((5 + 10) * 3 ** 2)                    # 135

# Example 2: Nested parentheses
print(((5 + 5) * 2) - (3 ** 2))             # 11

# Example 3: Multi-level nesting
print((((2 + 3) * 4) ** 2 - (10 // 3)) % 7) # 5

# Left-to-right evaluation
print(10 - 4 + 2)                           # 8

# Exponentiation is right-associative
print(2 ** 3 ** 2)                          # 512
```

### Data Types

Data types determine the kind of operations that can be performed on values. Each value in Python belongs to exactly one data type.

**Common Data Types:**

| Data Type | Examples | Definition |
|---|---|---|
| Integers | -2, -1, 0, 1, 2, 3 | Whole numbers, including negatives |
| Float | -1.0, -0.25, 0.0, 0.25 | Numbers with a decimal point |
| Strings | 'aa', 'Hello!', '11 cats' | Sequence of characters surrounded by quotes (' or ") |

**Note:** Even though 42 is an integer, 42.0 is a floating-point number.

**Note:** These are the three basic types covered so far. Python has more, such as booleans, lists, tuples, dictionaries, and sets, which come later in the course.

```python
# Integer examples
print(10 - 5)            # 5
print(7 * 3)             # 21
print(-8 + 12)           # 4

# Float examples
print(33.9 * 15)         # 508.5
print(5.5 / 2)           # 2.75
print(-4.2 + 3.1)        # -1.1

# String examples
print('Hello world!')    # Hello world!
print("Python is fun!")  # Python is fun!
print('1' * 5)           # 11111

# Integer vs float
print(type(42), type(42.0), type('42'))
# <class 'int'> <class 'float'> <class 'str'>
```

### Strings

In Python, a string (or `str`) is a sequence of characters surrounded by single (`'`) or double (`"`) quotes. Strings can include letters, numbers, symbols, and spaces.

Strings can also be multi-line by using triple quotes (`'''` or `"""`).

**String Concatenation**

The `+` operator joins two strings into a single, new string.

If you use `+` on a string and an integer, Python doesn't know how to handle it and raises a `TypeError`. To fix this, convert the integer to a string before concatenation.

**String Replication**

The `*` operator is normally used for multiplication, but with a string and an integer it becomes the **string replication operator**. It evaluates to a single string that repeats the original string as many times as the integer. Replication is useful, but less common than concatenation.

`*` can only be used with:
- two numeric values (multiplication), or
- one string and one integer (string replication)

Anything else raises an error. You can't multiply two strings, and you can't replicate a string a fractional number of times.

```python
# Strings with single and double quotes
print("Hello, World!")
print('To be, or not to be.')

# Multi-line string
print('''This is a string
that spans multiple lines.''')

# Concatenation
print('ai' + 'adventures')   # aiadventures

# Replication
print('ai' * 5)              # aiaiaiaiai

# Fix for str + int: convert the integer to a string
print('ai' + str(42))        # ai42
```

### Variables

A variable in Python is like an envelope in the computer's memory where you can store a single value. If you want to reuse the result of an evaluated expression later in your program, you can save it inside a variable.

**Assignment Statements**

You store values in variables using an assignment statement. This consists of a variable name, an equal sign (`=`), and the value to be stored.

Here, a variable named `var` is created with the value `42` stored in it. Think of a variable as a labeled envelope where the value is placed.

**Variables in Action**

Once a value is stored in a variable, you can use the variable in operations. You don't need to change the value `42` in every instance of your code. Simply changing the value assigned to `var` will update all the operations where the variable is used.

**Variables in Expressions**

Let's use some variables in expressions.

**Naming Variables in Python**

When naming variables in Python, follow a few basic rules to avoid errors and maintain readability.

**Rules for Naming Variables**

- **Single Word:** The variable name must be a single word.
- **Allowed Characters:** Only letters, numbers, and underscores (`_`) can be used.
- **No Starting with a Number:** The variable name cannot begin with a digit.

**Examples of Valid Names**

- `firstname`
- `firstName`
- `first_name`
- `_firstname`
- `FIRSTNAME`
- `first1`

These names follow the rules: no spaces, only valid characters, and not starting with a number.

**Important Points to Remember**

- **Case Sensitivity:** Python is case-sensitive, so `var`, `VAR`, `Var`, and `vAr` are treated as different variables.
- **Naming Convention:** It is common practice in Python to start variable names with a lowercase letter. This enhances readability and maintains a professional code style.
- **Use Descriptive Names:** Descriptive names clarify the purpose of each variable. For example, instead of `stuff`, consider `moving_boxes` or `school_supplies`.

**Understanding `print()`**

The `print()` function displays text on the screen. Quotes mark the boundaries of your string, but won't appear in the output. You can also print a blank line with `print()`.

**Note:** In Jupyter Notebook or Google Colab, you don't always need `print()`. The last line of code in a cell automatically displays the result. However, if you want to see intermediate results, you must use `print()`.

```python
# Assignment statement
var = 42
```

### Variable Overwriting

The term overwriting might sound familiar from file operations, where new content replaces old content. The same concept applies to variables: when a variable is reassigned, its previous value is replaced by the new one.

**Example 1: Numerical Overwriting**

At the end of the code block below, `var` is evaluated as 44 because the original value of 42 is replaced. The initial value is "forgotten" once the variable is overwritten.

**Example 2: String Overwriting**

Similarly, we can overwrite variables that store strings.

**Overwriting Across Multiple Cells**

You can overwrite a variable across multiple code cells in environments like Jupyter Notebooks. Each time the variable is reassigned, its old value is replaced and no longer accessible unless stored in a separate variable.

**Key Takeaway**

Overwriting allows you to reuse variables without keeping old data in memory, which is helpful in simplifying code and reducing clutter. You can replace a variable's value at any time, making it highly flexible for calculations, string manipulation, and more.

```python
# Example 1: Numerical overwriting
var = 42
var = var + 2  # Add 2 to the current value of 'var'
var            # Result: 44
```

```python
# Example 2: String overwriting
var = 'Hello'    # Initially set to 'Hello'
print(var)       # Output: Hello

var = 'Goodbye'  # Reassigning the value to 'Goodbye'
print(var)       # Output: Goodbye
```

```python
# Overwriting across multiple cells (run these two cells in order)
count = 10
```

```python
# Overwriting across multiple cells (run these two cells in order)
count = 10
```

```python
count = 20
print(count)     # Output: 20 (the old value 10 is gone)
```

**Understanding Comments in Python**

Comments are notes you leave in your code to explain what's happening. Python completely ignores them, so they don't affect how your program runs. Anything after the `#` is a comment. You can write comments on their own line or at the end of a code line.

**Why Use Comments?**

- **Explain your code:** comments remind you why you wrote a specific piece of code, especially after a long break.
- **Facilitate collaboration:** when others read your code, comments help them understand your logic and intentions.
- **Troubleshooting:** you can "turn off" code by adding a `#` in front of it.

**Common Error When Using `print()`**

Using `print()` with incompatible types can cause an error. This error occurs because you're attempting to concatenate a string (`'I am '`) with an integer (`24`).

**Solution:** convert the integer to a string using `str()`.

**Application in a Program**

This example combines `int()`, `str()`, and `input()` to handle user input and perform calculations.

If the user enters `'4'` for `myAge`:

1. `int(myAge)` converts `'4'` to `4`.
2. Adding 1 results in `5`.
3. `str(5)` converts `5` back to `'5'`, creating the final message: `'You will be 5 in a year.'`

**Text and Number Equivalence**

Python distinguishes between text and numbers, which affects equality comparisons.

1. **String and number comparison:** a string containing a number is not equal to an actual number (integer or float).
2. **Integer and float comparison:** an integer can be equal to a float if they represent the same value.
3. **Float representations:** leading zeros or extra decimal places do not change the numeric value, so different notations of the same number are considered equal.
