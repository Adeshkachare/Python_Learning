# Introduction to Python

### **What is Python**
Python is a high-level, general-purpose programming language known for its simple and readable syntax.

It is called a high-level language because its syntax is closer to human language than machine instructions, making it easier to develop and maintain programs.

It is commonly called an interpreted language because Python code is executed by a runtime environment, although CPython first compiles source code into bytecode.

### **What is an Interpreter?**
An interpreter is a software program that executes instructions written in a programming language.

It processes the code and helps the computer perform the operations the programmer intended.

Real-life example

Imagine you speak Marathi, but a person understands only English.
- You: Speak in Marathi.
- Interpreter: Understands your message and communicates it in English.
- Listener: Understands the message and responds.

Similarly, a programming language interpreter processes instructions written in a programming language so they can be executed by the computer.

The analogy isn't exact, though: a Python interpreter doesn't translate every Python statement directly into machine code one by one.

### **How does python interpreter work?**
image.png


The Python interpreter takes Python source code as input. In CPython, the source code is parsed and compiled into bytecode.

The Python virtual machine (PVM) then executes this bytecode using the CPython runtime.

The CPU executes the native machine instructions that implement the runtime and its operations.

**One extra detail:** CPython can reuse cached bytecode from .pyc files when appropriate, so it doesn't necessarily recompile every source file on every run.

---

### **What is python bytecode, and how it is different from machine code?**

Python bytecode is an intermediate set of instructions generated by the Python compiler from Python source code. These instructions are designed to be executed by the Python Virtual Machine (PVM), not directly by the CPU.

Difference between Python bytecode and machine code:
- Python bytecode: Platform-independent in concept, but its exact instructions depend on the Python implementation and version. The PVM executes these instructions.
- Machine code: Native binary instructions that a CPU can execute directly. They depend on the processor architecture, such as x86-64 or ARM.
Example:
- Python bytecode instructions: LOAD_CONST, BINARY_OP
- Machine code: Binary instruction patterns that the processor understands, such as encoded instructions for arithmetic operations.

**Execution flow in CPython:**
Python source code (.py) → Compiler → Python bytecode → PVM executes bytecode → CPU executes native machine instructions.

### **What is the difference between a compiler and an interpreter?**
A compiler and an interpreter are software components that help execute programs written in a programming language, but they process code differently.

**Compiler:**
A compiler translates the entire source code into another form, often native machine code, before the program executes. The generated machine code can be executed directly by a compatible CPU. Compilation may take some time initially, but the resulting program can execute efficiently. For example, a C compiler can convert C source code into a native executable.

**Interpreter:**
An interpreter executes a program through a runtime rather than necessarily generating a complete native machine-code executable beforehand. For example, in CPython, the source code is compiled into bytecode, which the Python Virtual Machine (PVM) executes. Python is commonly called an interpreted language because of this execution model, although it also involves compilation.

Example

C language:
C source code → Compiler → Native machine code → CPU executes it

Python (CPython):
Python source code → Compiler → Bytecode → Python Virtual Machine executes it

### **What is the difference between Python and other programming languages like C and Java?**

**Python:**

**Other programming languages:** Compared with C and Java, Python generally requires less code to perform the same task. ### **What are other implementations of Python, and how is “Python” different from “CPython”?**

Python is a language specification: a set of rules for how code is written and what it should do. CPython is one implementation of that specification, written in C. It is the default one you get from python.org, which is why people often use “Python” and “CPython” interchangeably.

Other implementations follow the same language but execute code differently:

PyPy: Uses a just-in-time (JIT) compiler, so long-running programs are often much faster than on CPython.
Jython: Runs on the Java Virtual Machine and can use Java libraries.
IronPython: Runs on .NET.

So the same .py source can be run by different implementations, and details like bytecode format or performance can vary between them.

### **What is the GIL (Global Interpreter Lock)?**

The GIL is a lock in CPython that allows only one thread to execute Python bytecode at a time, even on a multi-core CPU. It exists mainly to keep CPython’s memory management (reference counting) simple and safe.

Effects:

CPU-heavy tasks gain little from multiple threads, so use multiprocessing (separate processes, each with its own GIL) instead.
I/O-bound tasks (network calls, file reads) still benefit from threads, because the lock is released while waiting.

Extra detail: Recent CPython versions have been adding an experimental free-threaded build that removes the GIL. The default build still has it, so treat that as something to check against current docs.

### **What is __pycache__, and what are .pyc files?**

When CPython compiles a module to bytecode, it can save the result as a .pyc file inside a folder named __pycache__, next to the source file. For example, utils.py may produce __pycache__/utils.cpython-312.pyc.

Why it matters:

On the next run, Python can load the cached bytecode instead of recompiling, which speeds up startup (loading imports), not the running of the code itself.
The filename includes the Python version (cpython-312), because bytecode differs between versions.
Python checks whether the source file changed and recompiles if needed.
Only imported modules get cached. The main script you run directly is compiled each time.
It is safe to delete __pycache__, and it is normally excluded from version control via .gitignore.

### **What are pip and PyPI?**


PyPI (Python Package Index): The official public repository of third-party Python packages, such as requests, numpy, and pandas.
pip: The standard tool for installing packages from PyPI.

Example:

pip install requests

This downloads the requests package and its dependencies and makes it importable in your code.

Related commands:

pip list shows installed packages.
pip uninstall requests removes a package.
pip freeze > requirements.txt saves the installed versions, so others can reproduce your setup with pip install -r requirements.txt.

Good practice: Install packages inside a virtual environment (python -m venv venv) so each project keeps its own dependencies and they don’t conflict.

### **What is the difference between dynamic typing and static typing?**

In dynamic typing, which Python uses, a variable has no fixed type. The type belongs to the value, and types are checked while the program runs.

python
x = 10        # int
x = "hello"   # now a str, which is perfectly valid

In static typing (C, Java), you declare a variable’s type up front, and the compiler checks it before the program runs.

c
int x = 10;
x = "hello";  // compile-time error

Trade-off: Dynamic typing means less code and faster development, but some type errors only show up at runtime. Python offers optional type hints (x: int = 10) that tools like mypy can check without changing how the program runs.
