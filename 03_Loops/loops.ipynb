# Loops


## 1 Introduction

### Overview
Loops allow code to perform repetitive tasks without rewriting the same instructions again and again. A loop keeps executing a block of code until a specific condition is no longer met, like a favorite song on repeat.

Python has two main types of loops:
- `while` loop
- `for` loop

### Benefits of Loops
| Benefit | Description |
|---------|-------------|
| Repeat work easily | Loops run the same code again and again without writing it many times. |
| Save time | Programs become shorter by removing repeated steps. |
| Repeat actions | The same action can be done many times, like printing numbers. |
| Update screens | Loops keep screens or clocks updated by running code repeatedly. |

### `if` Statement vs `while` Loop
An `if` statement checks a condition and runs its block **only once** if the condition is `True`.

````python
i = 0            # initialization

if i <= 5:       # condition
    print(i)
    i = i + 1    # increment
````

**Output:**

```
0
```

A `while` loop keeps running **again and again** until the condition becomes `False`.

````python
i = 0            # initialization

while i <= 5:    # condition
    print(i)
    i = i + 1    # increment
````

## 2 While Loop

### Overview
A `while` loop executes a block of code repeatedly as long as a specified condition remains `True`. After each execution, the condition is checked again. When the condition becomes `False`, the loop stops.

### Components of a `while` Loop
- **The `while` keyword:** signals the start of the loop.
- **A condition:** an expression that evaluates to `True` or `False`.
- **A colon (`:`):** indicates that the following indented block is the loop body.
- **An indented block of code (the while clause):** the statements executed in each iteration.
- **(Optional) An increment/decrement statement:** updates a variable so the condition eventually becomes `False`.

### How a `while` Loop Works
1. **Condition evaluation:** at the start of each iteration, Python checks the condition.
2. **Execution:** if the condition is `True`, the loop body is executed.
3. **Update:** a variable is usually incremented or decremented so the condition can eventually become `False`.
4. **Repetition:** Python goes back and checks the condition again.
5. **Termination:** when the condition is `False`, the loop stops and the program continues with the rest of the code.

> **Important:** If the variable is never updated (no increment or decrement), the condition may stay `True` forever, creating an **infinite loop**.

### More Examples

#### Example 1: Print Even Numbers (2 to 10)
This loop starts at `i = 2` and prints even numbers up to 10, increasing by 2 each time.

````python
i = 2

while i <= 10:
    print(i)
    i += 2
````

### Example 2: Print Numbers Divisible by 5 (Ascending Order)
This loop prints the numbers divisible by 5 from 10 to 20.

````python
x = 10                # Initialize x at 10

while x <= 20:        # Continue as long as x is less than or equal to 20
    if x % 5 == 0:    # Check if x is divisible by 5
        print(x)
    x += 1            # Increment x by 1
````

### Example 3: Reverse Order
This loop prints the numbers divisible by 5 from 20 down to 10.

````python
x = 20                # Initialize x at 20

while x >= 10:        # Continue as long as x is greater than or equal to 10
    if x % 5 == 0:    # Check if x is divisible by 5
        print(x)
    x -= 1            # Decrement x by 1
````

## 3 For Loop

### Overview
A `for` loop repeats a block of code for each item in a sequence. It is useful when you know exactly how many times a block of code should run.

### Example 1: Iterating Over `"hello"`
- `word = "hello"` stores the string `"hello"`.
- `for char in word` loops through each character.
- `print(char)` prints the current character.

````python
word = "hello"

for char in word:
    print(char)
````

### Understanding the `range()` Function
General syntax:

```
range(start, stop, step)
```

| Parameter | Meaning |
|-----------|---------|
| `start` | First number in the sequence (defaults to `0` if omitted) |
| `stop` | One past the last number (exclusive) |
| `step` | Difference between consecutive numbers (defaults to `1`) |

#### Default Start
`range(6)` gives `0, 1, 2, 3, 4, 5`.

> `range()` by itself does not print anything. To see its values, convert it to a list or use it in a loop.

````python
print(range(6))          # shows the range object, not the numbers
print(list(range(6)))    # [0, 1, 2, 3, 4, 5]
````

### Custom Step
A step can be added along with a start and stop value.

````python
print(list(range(0, 10, 2)))   # [0, 2, 4, 6, 8]
````

### Custom Start and Step

````python
print(list(range(1, 10, 2)))   # [1, 3, 5, 7, 9]
````

### Using `range()` with a `for` Loop
`range()` is often paired with a `for` loop to generate a sequence of numbers. The stop value is **exclusive**, so the loop ends one number before it.

#### Basic Example: `range(1, 5)`

````python
for i in range(1, 5):
    print(i)
````

### Using a `for` Loop with Conditional Statements
Conditions can be added inside a loop, for example to print only multiples of 5.

````python
for x in range(10, 21):
    if x % 5 == 0:
        print(x)
````

## 4 Break & Continue

### Overview
The `break` and `continue` statements control how a loop runs. They work in both `for` loops and `while` loops.

### The `break` Statement
When `break` is encountered, the loop exits immediately and all remaining iterations are skipped.

#### Without `break`
Prints numbers from 1 to 9.

````python
for i in range(1, 10):
    print(i)
````

### With `break`
Stops when `i` equals 5.

````python
for i in range(1, 10):
    if i == 5:
        break
    print(i)
````

### `break` in a `while` Loop

````python
i = 1

while i < 10:
    if i == 5:
        break
    print(i)
    i += 1
````

## The `continue` Statement
When `continue` is encountered, the loop skips the rest of the current iteration and moves on to the next one. Unlike `break`, it does not end the loop.

### Without `continue`
Prints all numbers from 1 to 9 without skipping any value.

````python
for i in range(1, 10):
    print(i)
````

### With `continue`
Skips printing the number 5 and continues with the rest.

````python
for i in range(1, 10):
    if i == 5:
        continue
    print(i)
````

### `continue` in a `while` Loop
The increment must come **before** `continue`. Otherwise `i` never changes and the loop runs forever.

````python
i = 0

while i < 9:
    i += 1
    if i == 5:
        continue
    print(i)
````

## The `end` Argument in `print()`
The `end` argument controls what is printed after each item. By default, `print()` ends with a newline (`\n`), but this can be customized with `end=" "` (space), `end=""` (nothing), `end="-"`, or any other string.

### Without `end` (default)
Prints each number on a new line.

````python
for i in range(1, 4):
    print(i)
````

### With `end`
Prints all numbers on the same line, using different separators.

````python
print("Using end=' ':")
for i in range(1, 4):
    print(i, end=" ")      # Output: 1 2 3
print()

print("Using end='':")
for i in range(1, 4):
    print(i, end="")       # Output: 123
print()

print("Using end='-':")
for i in range(1, 4):
    print(i, end="-")      # Output: 1-2-3-
print()
````

### When to Use `end`
- To print results on the same line in loops or sequences.
- To format output (for example, space-separated or comma-separated values).
- To control spacing, punctuation, or layout in console output.

## When to Use Each Loop
| Loop | Use when | Example |
|------|----------|---------|
| `for` | The exact number of iterations is known | Iterating through a range or a list |
| `while` | Looping until a condition changes | Asking the user again until the input is valid |

## Additional Notes
- **Nested loops:** `break` and `continue` affect only the innermost loop.
- **Homework:** Search for the `end` argument of `print()` in Python to learn how to customize line endings.

## 5 Further Loops

### 1. Nested Loops

#### Definition
In programming, **nested** means one structure placed inside another. **Nested loops** are a loop inside another loop. Examples include:
- a `while` loop inside another `while` loop
- a `for` loop inside another `for` loop
- a mix of `for` and `while` loops

#### How They Work
The **inner loop** completes all of its iterations before the **outer loop** moves to its next step. This is useful for creating combinations or grids of numbers.

#### Example (Using `for` Loops)

````python
for i in range(3):
    for j in range(3):
        print(i, j)
````

### Task
Try achieving the same output using `while` loops.

````python
i = 0

while i < 3:
    j = 0

    while j < 3:
        print(i, j)
        j += 1

    i += 1
````

### 2. Infinite Loop

#### Definition
An **infinite loop** is a loop that continues endlessly until it is stopped manually or a `break` statement is used. Infinite loops often happen unintentionally, especially in `while` loops that never reach a stopping condition.

#### Example
The loop below never ends because `i` is never updated, so the condition `i <= 5` is always `True`.

```
i = 0

while i <= 5:
    print(i)
```

> **Note:** In environments like Google Colab, interrupt the runtime to stop an infinite loop (click the stop button on the cell, or use **Runtime → Interrupt execution**).

> **Tip:** Infinite loops can be useful in continuous-operation projects, like the classic "3 Missionaries and 3 Cannibals" problem.

#### Fixing the Infinite Loop
Adding an increment lets the condition eventually become `False`.

````python
i = 0

while i <= 5:
    print(i)
    i += 1     # increment makes the loop stop
````

## Q1. How is the `while` loop different from the `for` loop?

| `while` loop | `for` loop |
|--------------|------------|
| Repeats as long as a condition is `True` | Repeats for each item in a sequence |
| Used when the number of iterations is not known in advance | Used when the number of iterations is known |
| The loop variable must be updated manually | The loop variable is updated automatically |
| Can become an infinite loop if the condition never becomes `False` | Ends when the sequence is finished |

````python
# while loop: runs until the condition becomes False
i = 1
while i <= 3:
    print("while:", i)
    i += 1

# for loop: runs once for each item in the sequence
for i in range(1, 4):
    print("for:", i)
````

## Q2. Define the `range()` function in Python and its arguments.

`range()` generates a sequence of integers. Syntax:

```
range(start, stop, step)
```

| Argument | Meaning |
|----------|---------|
| `start` | First number in the sequence (default `0`) |
| `stop` | One past the last number (exclusive, required) |
| `step` | Difference between consecutive numbers (default `1`) |

````python
print(list(range(5)))          # [0, 1, 2, 3, 4]
print(list(range(2, 8)))       # [2, 3, 4, 5, 6, 7]
print(list(range(0, 10, 3)))   # [0, 3, 6, 9]
print(list(range(5, 0, -1)))   # [5, 4, 3, 2, 1]
````

## Q3. How do you use `break` and `continue` statements in a loop?

- `break` exits the loop immediately, skipping all remaining iterations.
- `continue` skips the rest of the current iteration and moves to the next one.

Both are usually placed inside an `if` statement. In nested loops, they affect only the innermost loop.

````python
for i in range(1, 8):
    if i == 3:
        continue      # skip 3
    if i == 6:
        break         # stop the loop at 6
    print(i)          # prints 1, 2, 4, 5
````

## Q4. Which loop is used to create an infinite loop?

The `while` loop. Writing `while True:` creates a loop that never ends on its own, so a `break` statement is normally used to exit it. (A `while` loop can also become infinite by mistake if its condition never becomes `False`.)

````python
# Infinite loop that is ended with break
count = 0

while True:
    print("count:", count)
    count += 1
    if count == 3:
        break
````

## Q5. Can we use float values in the step argument of `range()`?

No. `range()` accepts only integers for `start`, `stop`, and `step`. Passing a float raises a `TypeError`.

```
range(0, 1, 0.1)   # TypeError: 'float' object cannot be interpreted as an integer
```

To step through floats, use a `while` loop.

````python
# Float steps using a while loop
x = 0.0

while x < 1:
    print(round(x, 1))
    x += 0.25
````
