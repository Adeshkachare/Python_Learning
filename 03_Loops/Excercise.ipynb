## Create a Program to Display the First 10 Even Numbers in Reverse Order

Create a program to display the first 10 even numbers in reverse order in a single line using the `end` parameter of the `print` function.

**Note:** Solve this using both `for` and `while` loops.

````python
for i in range(10, -1, -2):
  print(i, end= ' ')
````

## Create a Program to Display Numbers Starting from 20 to 40 with a Difference of 5

Create a program that displays numbers starting from 20 to 40 with a difference of 5 in a single line using the `end` parameter of the `print` function.

**Note:** Solve this using both `for` and `while` loops.

````python
i = 20
while i <= 40:
  print(i)
  i += 5
````

````python
for i in range(20,41,5):
  print(i)
````

## Create a Program to Display the Cube of the First 5 Numbers

Create a program to display the cube of the first 5 numbers in a single line using the `end` parameter of the `print` function.

**Note:** Solve this using both `for` and `while` loops.

````python
for i in range(1, 6):
  print(i**3)
````

````python
i = 1
while i < 6:
  print(i**3)
  i += 1
````

## Create a Program to Display the Factorial of a Number

Create a program to display the factorial of a number.

**Note:** Use both loops with positive and negative step values.

````python
n = int(input("Enter a number: "))
fact = 1
for i in range(1,n+1):
  fact = fact * i
  print(fact)
````

## Create a Program to Accept Input from the User Until 0 is Entered

Create a program that keeps accepting input from the user until `0` is entered and prints the sum of all the numbers entered at the end.

````python
n = None
count = 0
while n != 0:
  n = int(input("Enter a number: "))
  count += n
print(count)
````

## Create a Program to Print a Series of Numbers

Create a program to print the following series for `n` terms, where `n` is taken from the user:

**Series:** 2, 22, 222, 2222, ... up to `n` terms

````python
n = int(input("Enter a number"))
for i in range(1,n+1):
  print('2'*i,end=' ')
````

## Create a Program to Display the First Number Divisible by 5

Create a program that displays the first number divisible by 5 between 6 and 16, then `break` the loop.

**Note:** Solve this using both `for` and `while` loops.

````python
for i in range(1,500):
  if i%5 == 0 and i%6 == 0 and i%16 == 0:
    print(i)
    break
````

## Create a Program to Print Multiplication Tables

Create a program that replicates the following output:

```
The Multiplication Table of 2
2 x 1 = 2
2 x 2 = 4
2 x 3 = 6
...
2 x 10 = 20

The Multiplication Table of 3
3 x 1 = 3
3 x 2 = 6
...
3 x 10 = 30

The Multiplication Table of 4
4 x 1 = 4
4 x 2 = 8
...
4 x 10 = 40

Happy Learning!
```

**Note:** Solve this using both `for` and `while` loops.

````python
for i in range(1,11):
  for j in range(1,11):
    print(f"{i} * {j} == {i*j}")
  print()
````
