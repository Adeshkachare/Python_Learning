## 1 Introduction to Flow Control

### Overview
- By default, code runs from top to bottom, line by line.
- Flow control lets a program decide, skip, or repeat instructions based on conditions.
- This makes programs dynamic and responsive.

### Flowchart Symbols
| Shape | Meaning |
|-------|---------|
| Oval | Start / End |
| Rectangle | Step |
| Diamond | Yes/No decision |

### Example: Alarm Clock
- Does the alarm ring? No → go back to sleep.
- Does the alarm ring? Yes → is it time to get up?
  - Yes → get out of bed.
  - No → hit snooze and wait for the next alarm.

````python
# Alarm flowchart as code
alarm_rings = True
time_to_get_up = False

if alarm_rings:
    if time_to_get_up:
        print("Get out of bed")
    else:
        print("Hit snooze")
else:
    print("Go back to sleep")
````

## 3 Comparison Operators

### Overview
Comparison operators compare two values and return a Boolean (`True` or `False`).

| Operator | Meaning |
|----------|---------|
| `a == b` | equal to |
| `a != b` | not equal to |
| `a < b` | less than |
| `a <= b` | less than or equal to |
| `a > b` | greater than |
| `a >= b` | greater than or equal to |

### `==` vs `=`
- `=` assigns a value to a variable.
- `==` checks whether two values are equal.

````python
a = 6
b = 4

print(a == b)   # False
print(a != b)   # True
print(a > b)    # True
print(a >= b)   # True
print(a < b)    # False
print(a <= b)   # False
````

### Booleans and Integers
- `True` is equivalent to `1`, and `False` is equivalent to `0`.
- `bool(0)` is `False`; `bool()` of any other number is `True`.
- Dividing by `False` (which equals 0) raises a `ZeroDivisionError`.

````python
print(True == 1)     # True
print(False == 0)    # True

print(bool(1), bool(0))   # True False
print(bool(a), bool(b))   # True True

print(a + True)      # 7
print(b * False)     # 0
print(a / True)      # 6.0

# print(b // False)  # ZeroDivisionError (remove # to test)
````

## 4 Boolean Operators

### Overview
Boolean operators combine Boolean values and return a Boolean result.

| Operator | Result |
|----------|--------|
| `and` | `True` only if both values are `True` |
| `or` | `True` if at least one value is `True` |
| `not` | Returns the opposite value |

### Truth Tables
| A | B | A and B | A or B |
|---|---|---------|--------|
| False | False | False | False |
| False | True | False | True |
| True | False | False | True |
| True | True | True | True |

| A | not A |
|---|-------|
| False | True |
| True | False |

````python
# and
print(False and False)   # False
print(False and True)    # False
print(True and False)    # False
print(True and True)     # True

# or
print(False or False)    # False
print(False or True)     # True
print(True or False)     # True
print(True or True)      # True

# not
print(not True)          # False
print(not False)         # True
print(not (not True))    # True
````

### Combining Operators
- Parentheses are evaluated first.
- Precedence (highest to lowest): `not` → `and` → `or`

````python
print(True and (False or True))             # True
print((not False) and (True or False))      # True

# Precedence without parentheses
print(True or False and False)              # True  (and runs first)
print((True or False) and False)            # False (parentheses change the order)
````

### Short-Circuit Evaluation
- `and`: if the first value is `False`, Python skips the second.
- `or`: if the first value is `True`, Python skips the second.

Boolean operators are widely used in `if` statements, loops, and data filtering.

````python
# The second part is never evaluated, so no error occurs
print(False and (1 / 0))   # False
print(True or (1 / 0))     # True
````

## 5 Comparison & Boolean Operators Together

### Overview
Comparison operators evaluate to `True` or `False`, so they can be combined with `and`, `or`, and `not` to build complex logical expressions.

### How Python Evaluates an Expression
1. Evaluate the left expression.
2. Evaluate the right expression.
3. Combine the results into a single Boolean value.

Example: `(1 < 2) and (3 < 4)` → `True and True` → `True`

````python
print((1 < 2) and (3 < 4))     # True  (True and True)
print((4 < 5) and (7 < 6))     # False (True and False)
print((1 == 2) or (2 == 2))    # True  (False or True)
````

````python
print(((2 + 2) == 4) and (not (2 + 2 == 5)) and ((2 * 2) == (2 + 2)))   # True
````

## 6 Flow Control Statements

### Overview
Flow control statements let a program run specific blocks of code based on conditions. In a flowchart, they appear as diamonds, where the path depends on whether the condition is `True` or `False`.

### 1. `if` Statement
Runs a block of code only if the condition is `True`.

```
if condition:
    # block of code (if clause)
```

````python
grade = 93

if grade >= 90:
    print('Grade A')
````

### 2. `else` Statement
Runs a block of code when the preceding `if` condition is `False`.

```
if condition:
    # block of code (if clause)
else:
    # block of code (else clause)
```

````python
grade = 88

if grade >= 90:
    print('Grade A')
else:
    print('Not Grade A')
````

### 3. `elif` Statement
Checks multiple conditions in sequence. Python runs the first block whose condition is `True`, and `else` runs only if none of the conditions are `True`.

```
if condition1:
    # block of code (if clause)
elif condition2:
    # block of code (elif clause)
else:
    # block of code (else clause)
```

````python
grade = 75

if grade >= 90:
    print('Grade A')
elif grade >= 80:
    print('Grade B')
else:
    print('Grade C')
````

## 7 Elements of Flow Control

### Overview
Flow control guides the execution of code based on certain conditions. It decides what to do next based on the situation. In Python it revolves around two concepts: **conditions** and **blocks of code**.

### 1. Conditions
A condition is a Boolean expression that evaluates to `True` or `False`. It determines which path the program takes.

- Written using comparison operators such as `<`, `>`, `==`, `!=`.
- Used in `if`, `elif`, and `else` statements.

````python
a = 5
b = 10

# Examples of conditions
condition1 = (a < b)    # True: 5 is less than 10
condition2 = (a == b)   # False: 5 is not equal to 10

print(condition1)
print(condition2)
````

### 2. Blocks of Code
A block is a group of one or more lines of code that belong together. Blocks define the code that runs under a specific condition and organize logic into readable chunks.

### Indentation
Python uses indentation (spaces or tabs at the start of a line) to define blocks.

**Rules:**
- A block begins after a colon (`:`) and requires an indented line.
- Blocks can be nested (a block inside another block).
- A block ends when the indentation level decreases.

#### Basic Example

````python
x = 7

if x < 10:
    print('Less than 10')   # indented block
````

#### Nested Blocks

````python
x = 7

if x < 10:
    print('Less than 10')               # Block 1
    if x > 5:
        print('Also greater than 5')    # Nested block
````

### 3. Common Error: IndentationError
Improper indentation raises an `IndentationError`, one of the most common Python errors.

**Incorrect code:**

```
x = 7

if x < 10:
print('Less than 10')   # missing indentation
```

**Error output:**

```
IndentationError: expected an indented block
```

**Fix:** indent the line inside the block (see below).

````python
x = 7

if x < 10:
    print('Less than 10')   # correctly indented
````

## 8 Nested if Statements

### Overview
Nesting means placing one thing inside another, like Russian dolls or a lunchbox inside a backpack. A nested `if` is an `if` statement inside another `if` statement.

- Use it to check a second condition only if the first condition is `True`.
- It allows more detailed decisions: answer "yes" to one question, then ask a follow-up question.

### Example: Checking a Number

````python
x = 7

if x < 10:                      # First check
    print("Less than 10")
    if x > 5:                   # Nested check
        print("Greater than 5")
    else:
        print("Less than or equal to 5")
````

**Output (x = 7):**

```
Less than 10
Greater than 5
```

### Importance of Indentation
Indentation defines which block each line belongs to. Without correct indentation, Python raises an `IndentationError`.

**Incorrect code:**

```
x = 7

if x < 10:
print('Less than 10')   # this line should be indented
    if x > 5:
        print('Greater than 5')
    else:
        print('Less than or equal to 5')
```

**Error output:**

```
IndentationError: expected an indented block
```

### Key Points
- Nesting allows more complex decision structures.
- Indentation determines the scope of each block.
- Multiple `if` statements can be nested to create a hierarchy of conditions.
- Each `if` can have its own `else`, and `elif` blocks can also be nested.

### Practical Use Cases
- User input validation
- Multi-level decision-making
- Game mechanics where several states must be checked

````python
# Nested if with elif and else (login check)
username_correct = True
password_correct = False

if username_correct:
    if password_correct:
        print("Login successful")
    else:
        print("Wrong password")
else:
    print("Unknown username")
````

### Q1. List all the Boolean operators and their truth tables.

The Boolean operators are `and`, `or`, and `not`.

**`and`** (True only if both are True)

| A | B | A and B |
|---|---|---------|
| False | False | False |
| False | True | False |
| True | False | False |
| True | True | True |

**`or`** (True if at least one is True)

| A | B | A or B |
|---|---|--------|
| False | False | False |

**`not`** (returns the opposite)

| A | not A |
|---|-------|

````python
# Verifying the truth tables
for A in (False, True):
    for B in (False, True):
        print(A, B, "| and:", A and B, "| or:", A or B)

print()
for A in (False, True):
    print(A, "| not:", not A)
````

### Q2. What is the order of precedence of the Boolean operators?

From highest to lowest: **`not` → `and` → `or`**

````python
# and runs before or
print(True or False and False)     # True  -> True or (False and False)
print((True or False) and False)   # False -> parentheses change the order
````

### Q3. Does the `else` statement have a condition to check?

No. `else` has no condition. It runs only when the conditions of the preceding `if` (and any `elif`) are all `False`.

````python
x = 3

if x > 5:
    print("Greater than 5")
else:                         # no condition written here
    print("5 or less")
````

## Q4. What is the minimum and maximum number of `else` and `elif` statements in a program?

| Statement | Minimum | Maximum |
|-----------|---------|---------|
| `else` | 0 | 1 (per `if` chain) |
| `elif` | 0 | No fixed limit (per `if` chain) |

Both are optional. A single `if` chain can have only one `else`, and it must come last.

## Q5. Can we use `elif` before an `if` statement?

No. An `elif` must always come after an `if`. Using it first raises a `SyntaxError`.

```
elif x > 5:        # SyntaxError: invalid syntax
    print("Hello")
```

````python
# One if, many elifs, at most one else
marks = 65

if marks >= 90:
    print("Grade A")
elif marks >= 80:
    print("Grade B")
elif marks >= 70:
    print("Grade C")
elif marks >= 60:
    print("Grade D")
else:
    print("Fail")
````
