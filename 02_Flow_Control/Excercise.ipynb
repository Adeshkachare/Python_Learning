## Exercise 3: Bike Price Program with Tax Calculation
Create a program to accept the cost price of a bike and display the total price of the bike, including the tax to be paid, according to the following criteria.

| Cost Price (in Rs) | Tax Rate |
|--------------------|----------|
| > 100000 | 15% |
| > 50000 and <= 100000 | 10% |
| <= 50000 | 5% |

```python
price = int(input("Enter bike price: "))

if price > 100000:
  price += price * 0.15
  print(price)
elif price > 50000 and price <= 100000:
  price += price *0.10
  print(price)
else:
  price += price * 0.05
  print(price)
```

## Exercise 4: Attendance Checker for Exam Eligibility
A student must have an attendance of at least 75% to be allowed to sit for the exams. Create a program to check whether the student satisfies this criterion by taking 2 inputs from the user and displaying the necessary output message, along with the attendance percentage, using the `print` function.

**Hint:** Take user inputs for:
- The number of classes held
- The number of classes attended



```python
classes_held = int(input("Enter the number of classes held: "))
classes_attended = int(input("Enter the number of classes attended: "))

attendance = (classes_attended / classes_held) * 100

print("Attendance percentage:", attendance, "%")

if attendance >= 75:
    print("Eligible to sit for the examination")
else:
    print("Not eligible to sit for the examination")
```

## Exercise 6: Program to Check Leap Year
Create a program to check whether a given year is a leap year or not.

**Note:** Make sure you enter an appropriate message in the `input` function that assists the user in entering the values.

**Hint:** Search online for all the conditions a year must satisfy to be a leap year.

```python
year = int(input("Enter a year: "))

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(year, "is a Leap Year")
else:
    print(year, "is not a Leap Year")
```

## Exercise 7: Program for Library Charges
Create a program that accepts the number of days from the user and calculates the library charge according to the following table.

**Note:** Make sure you enter an appropriate message in the `input` function that assists the user in entering the values.

| Number of Days | Cost per Day |
|----------------|--------------|
| Till 5 days | Rs 2 / day |
| 6 to 10 days | Rs 3 / day |
| 11 to 15 days | Rs 4 / day |
| After 15 days | Rs 5 / day |

```python
days = int(input("Enter the number of days: "))

if days <= 5:
    charge = days * 2

elif days <= 10:
    charge = (5 * 2) + (days - 5) * 3

elif days <= 15:
    charge = (5 * 2) + (5 * 3) + (days - 10) * 4

else:
    charge = (5 * 2) + (5 * 3) + (5 * 4) + (days - 15) * 5

print("Total library charges: Rs", charge)
```
