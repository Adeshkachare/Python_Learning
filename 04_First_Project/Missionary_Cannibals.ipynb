The game must go on until you win or lose, so lets put all the conditions inside an infinite loop :

- Connect all the segments of the code according to the above structure and do the necessary amendments.
- If a move is invalid then without making any changes, you have to `continue`.

  This is done so that we get a second chance to change our inputs and start over from the same point without losing previous amendments.

- If you Win or Lose the game `break` the loop and get out of it.
- Check the winning and losing condition everytime after the boat has travelled.
- After getting out of the loop print `Game Over`.

````python

boat_side = 'Right'

missionaries_on_right = 3
cannibals_on_right = 3

missionaries_on_left = 0
cannibals_on_left = 0

print(f"M = {missionaries_on_left} C = {cannibals_on_left}  |-----B|  M = {missionaries_on_right} C = {cannibals_on_right}")

while True:

    missionaries = int(input("Enter the number of missionaries to put on the boat: "))
    cannibals = int(input("Enter the number of cannibals to put on the boat: "))

    # Invalid move 1: Boat capacity
    if missionaries + cannibals > 3 or missionaries + cannibals < 0:
        print("Invalid move 1")
        continue

    # Check boat position
    if boat_side == 'Right':
        print("The boat is on the right side")
    else:
        print("The boat is NOT on the right side")

    # Check available people on the right
    if boat_side == 'Right':
        if missionaries > missionaries_on_right or cannibals > cannibals_on_right:
            print("Invalid move 2")
            continue

    # Move people from right to left
    if boat_side == 'Right':
        missionaries_on_right = missionaries_on_right - missionaries
        cannibals_on_right = cannibals_on_right - cannibals

        missionaries_on_left = missionaries_on_left + missionaries
        cannibals_on_left = cannibals_on_left + cannibals

    # Update boat position
    if boat_side == 'Right':
        print(f"M = {missionaries_on_left} C = {cannibals_on_left}  |B-----|  M = {missionaries_on_right} C = {cannibals_on_right}")
        boat_side = 'Left'

    # Check boat position
    if boat_side == 'Left':
        print("The boat is on the left side")
    else:
        print("The boat is on the right side")

    # Check available people on the left
    if boat_side == 'Left':
        if missionaries > missionaries_on_left or cannibals > cannibals_on_left:
            # This check is only relevant before moving back.
            # The previous crossing has already happened, so we handle
            # the return crossing in the next iteration.
            pass

    # Move people from left to right on the next turn
    # (Handled by the boat_side condition below in the next iteration)

    # Check losing condition
    if missionaries_on_right < cannibals_on_right and missionaries_on_right > 0:
        print("YOU LOSE")
        break

    if missionaries_on_left < cannibals_on_left and missionaries_on_left > 0:
        print("YOU LOSE")
        break

    # Check winning condition
    if missionaries_on_left == 3 and cannibals_on_left == 3:
        print("YOU WIN")
        break

    # Return journey: next iteration handles the boat's current side
    if boat_side == 'Left':
        missionaries = int(input("Enter missionaries for return journey: "))
        cannibals = int(input("Enter cannibals for return journey: "))

        if missionaries + cannibals > 2 or missionaries + cannibals < 1 or missionaries < 0 or cannibals < 0:
            print("Invalid move 1")
            continue

        if missionaries > missionaries_on_left or cannibals > cannibals_on_left:
            print("Invalid move 2")
            continue

        missionaries_on_left -= missionaries
        cannibals_on_left -= cannibals

        missionaries_on_right += missionaries
        cannibals_on_right += cannibals

        print(f"M = {missionaries_on_left} C = {cannibals_on_left}  |-----B|  M = {missionaries_on_right} C = {cannibals_on_right}")
        boat_side = 'Right'

    # Check losing condition after return
    if ((missionaries_on_right < cannibals_on_right and missionaries_on_right > 0) or
        (missionaries_on_left < cannibals_on_left and missionaries_on_left > 0)):
        print("YOU LOSE")
        break

print("Game Over")
````

````python
# Missionaries & Cannibals - 3D + terminal UI for a single Google Colab cell
# Paste this whole cell into Colab and run it. (Needs internet for three.js.)

from IPython.display import HTML, display

GAME_HTML = r"""
<link href="https://fonts.googleapis.com/css2?family=VT323&display=swap" rel="stylesheet">
<style>
#mc{--g:#38ff7a;--dim:#1f8f4a;--amb:#ffb000;--red:#ff4d4d;--cy:#6fe3ff;
  font-family:'VT323','Courier New',monospace;font-size:19px;color:var(--g);background:#020a05;
  border:2px solid var(--dim);border-radius:10px;max-width:900px;margin:0 auto;position:relative;overflow:hidden;box-sizing:border-box}
#mc *{box-sizing:border-box}
#mc .bar{display:flex;justify-content:space-between;padding:6px 12px;border-bottom:1px solid var(--dim);background:#04170b;color:var(--g)}
#mc .bar b{font-weight:400;color:var(--amb)}
#mc #mc-view{height:340px;position:relative;border-bottom:1px solid var(--dim);cursor:grab;touch-action:none}
#mc canvas{display:block;width:100%;height:100%}
#mc .hud{position:absolute;left:10px;top:8px;font-size:17px;color:var(--g);text-shadow:0 0 6px var(--g);pointer-events:none;line-height:1.15}
#mc .hint{position:absolute;right:10px;bottom:6px;font-size:15px;color:var(--dim);pointer-events:none}
#mc #mc-log{height:230px;overflow-y:auto;padding:10px 12px;white-space:pre-wrap;word-break:break-word;line-height:1.2;text-shadow:0 0 5px rgba(56,255,122,.55)}
#mc .ok{color:var(--g)} #mc .err{color:var(--red)} #mc .warn{color:var(--amb)} #mc .dim{color:var(--dim)} #mc .cy{color:var(--cy)}
#mc .prompt{display:flex;gap:8px;padding:6px 12px 8px;border-top:1px dashed var(--dim)}
#mc .prompt span{color:var(--cy);white-space:nowrap}
#mc input{flex:1;min-width:0;background:transparent;border:0;outline:0;color:var(--g);font:inherit;caret-color:var(--g);text-shadow:0 0 5px var(--g)}
#mc .keys{display:flex;flex-wrap:wrap;gap:6px;padding:0 12px 12px}
#mc .keys button{font:inherit;color:var(--g);background:#04170b;border:1px solid var(--dim);border-radius:4px;padding:2px 10px;cursor:pointer}
#mc .keys button:hover{background:var(--g);color:#020a05}
#mc .keys button:focus-visible{outline:2px solid var(--amb)}
#mc .crt{position:absolute;inset:0;pointer-events:none;
  background:repeating-linear-gradient(to bottom,rgba(0,0,0,0) 0 2px,rgba(0,0,0,.22) 3px),radial-gradient(ellipse at center,rgba(0,0,0,0) 60%,rgba(0,0,0,.5) 100%)}
@media(max-width:560px){#mc{font-size:16px}#mc #mc-view{height:260px}}
</style>

<div id="mc">
  <div class="bar"><span>rescue@river:~ missionaries.exe</span><span>moves <b id="mc-mv">0</b> | boat <b id="mc-bs">RIGHT</b></span></div>
  <div id="mc-view">
    <div class="hud" id="mc-hud"></div>
    <div class="hint">drag to rotate</div>
  </div>
  <div id="mc-log" aria-live="polite"></div>
  <div class="prompt"><span>rescue@river:~$</span><input id="mc-in" autocomplete="off" spellcheck="false" aria-label="command"></div>
  <div class="keys">
    <button data-c="+m">+M</button><button data-c="+c">+C</button><button data-c="-m">-M</button><button data-c="-c">-C</button>
    <button data-c="go">GO</button><button data-c="undo">UNDO</button><button data-c="reset">RESET</button><button data-c="help">HELP</button>
  </div>
  <div class="crt"></div>
</div>

<script>
(function(){
const $=id=>document.getElementById(id);
const log=$('mc-log'), inp=$('mc-in');

/* ---------- terminal ---------- */
function say(t,cls){const d=document.createElement('div');if(cls)d.className=cls;d.textContent=t;log.appendChild(d);
  while(log.children.length>250)log.removeChild(log.firstChild);log.scrollTop=log.scrollHeight;}
const ok=t=>say('[ OK ] '+t,'ok'), er=t=>say('[ERR ] '+t,'err'), wn=t=>say('[WARN] '+t,'warn');

/* ---------- game state ---------- */
const chars=[]; let S, hist=[], busy=false, fx=null;
const sideName=s=>s==='L'?'left':'right';
const cnt=(loc,t)=>chars.filter(c=>c.loc===loc&&(!t||c.type===t)).length;
const unsafe=(m,c)=>m>0&&c>m;
function reset(){
  chars.forEach((c,i)=>c.loc='R'); S={boat:'R',moves:0,over:false}; hist=[]; busy=false; fx=null; layout(); paintBanks(); hud();
}
function hud(){
  const bl=chars.filter(c=>c.loc==='B');
  $('mc-mv').textContent=S.moves; $('mc-bs').textContent=S.boat==='L'?'LEFT':'RIGHT';
  const L='LEFT  M:'+cnt('L','M')+' C:'+cnt('L','C'), R='RIGHT M:'+cnt('R','M')+' C:'+cnt('R','C');
  $('mc-hud').textContent=L+'\n'+R+'\nBOAT  M:'+bl.filter(c=>c.type==='M').length+' C:'+bl.filter(c=>c.type==='C').length+' /2';
}
function status(){
  const b=S.boat==='L'?'[BOAT]~~~~~~~':'~~~~~~~[BOAT]';
  say('LEFT [M:'+cnt('L','M')+' C:'+cnt('L','C')+']  '+b+'  RIGHT [M:'+cnt('R','M')+' C:'+cnt('R','C')+']  boat load: M:'+
    chars.filter(c=>c.loc==='B'&&c.type==='M').length+' C:'+chars.filter(c=>c.loc==='B'&&c.type==='C').length,'cy');
}
function board(t){
  if(S.over||busy)return er(S.over?'Game over. Type reset or undo.':'Boat is sailing.');
  if(chars.filter(c=>c.loc==='B').length>=2)return er('Boat is full (capacity 2).');
  const c=chars.find(c=>c.type===t&&c.loc===S.boat);
  if(!c)return er('No '+(t==='M'?'missionary':'cannibal')+' left on the '+sideName(S.boat)+' bank.');
  c.loc='B'; ok((t==='M'?'Missionary':'Cannibal')+' boarded.'); layout(); hud();
}
function unboard(t){
  if(S.over||busy)return er(S.over?'Game over. Type reset or undo.':'Boat is sailing.');
  const c=chars.find(c=>c.type===t&&c.loc==='B');
  if(!c)return er('No '+(t==='M'?'missionary':'cannibal')+' on the boat.');
  c.loc=S.boat; ok((t==='M'?'Missionary':'Cannibal')+' left the boat.'); layout(); hud();
}
function load(m,c){
  if(S.over||busy)return er(S.over?'Game over. Type reset or undo.':'Boat is sailing.');
  if(m<0||c<0||m+c>2)return er('Invalid move: boat holds at most 2.');
  if(m>cnt(S.boat,'M')+chars.filter(x=>x.loc==='B'&&x.type==='M').length||c>cnt(S.boat,'C')+chars.filter(x=>x.loc==='B'&&x.type==='C').length)
    return er('Not enough people on the '+sideName(S.boat)+' bank.');
  chars.forEach(x=>{if(x.loc==='B')x.loc=S.boat;});
  for(let i=0;i<m;i++)chars.find(x=>x.type==='M'&&x.loc===S.boat).loc='B';
  for(let i=0;i<c;i++)chars.find(x=>x.type==='C'&&x.loc===S.boat).loc='B';
  ok('Boat loaded: M:'+m+' C:'+c); layout(); hud();
}
function go(){
  if(S.over)return er('Game over. Type reset or undo.');
  if(busy)return er('Boat is already sailing.');
  const bl=chars.filter(c=>c.loc==='B');
  if(!bl.length)return er('The boat cannot cross empty. Load someone first.');
  hist.push({locs:chars.map(c=>c.loc==='B'?S.boat:c.loc),boat:S.boat,moves:S.moves});
  const to=S.boat==='R'?'L':'R'; busy=true; S.boat=to; S.moves++; say('Crossing to the '+sideName(to)+' bank...','dim'); hud();
  setTimeout(()=>{
    bl.forEach(c=>c.loc=to); busy=false; layout(); hud();
    const bad=unsafe(cnt('L','M'),cnt('L','C'))||unsafe(cnt('R','M'),cnt('R','C'));
    if(bad){S.over=true;fx='lose';paintBanks();status();
      er('FAIL: cannibals outnumber missionaries on the '+(unsafe(cnt('L','M'),cnt('L','C'))?'left':'right')+' bank. YOU LOSE.');
      say("Type 'undo' to take the move back or 'reset' to restart.",'dim');return;}
    if(cnt('L','M')===3&&cnt('L','C')===3){S.over=true;fx='win';paintBanks();status();
      ok('All six crossed in '+S.moves+' moves (optimal: 11). YOU WIN.');return;}
    status();
  },1400);
}
function undo(){
  if(busy)return er('Wait for the boat to arrive.');
  const h=hist.pop(); if(!h)return er('Nothing to undo.');
  chars.forEach((c,i)=>c.loc=h.locs[i]); S.boat=h.boat; S.moves=h.moves; S.over=false; fx=null; layout(); paintBanks(); hud(); ok('Move undone.'); status();
}
function help(){
  [ 'COMMANDS',
    '  load <m> <c>   set boat load from the boat\'s bank, e.g. load 1 1',
    '  1 1            shortcut for load 1 1',
    '  +m +c          put a missionary / cannibal on the boat',
    '  -m -c          take one off the boat',
    '  go             cross the river',
    '  undo / reset   take back a move / restart',
    '  status         show the current state',
    'RULES  Move all 6 from RIGHT to LEFT. Boat holds 1 or 2.',
    '       If cannibals outnumber missionaries on a bank (and a missionary is there) you lose.'
  ].forEach(l=>say(l,'cy'));
}
function exec(line){
  line=line.trim(); if(!line)return; say('rescue@river:~$ '+line,'dim');
  const w=line.toLowerCase().split(/\s+/); const a=w[0]; let m;
  if(a==='help'||a==='?')help();
  else if(a==='go'||a==='cross')go();
  else if(a==='undo')undo();
  else if(a==='reset'||a==='restart'){reset();ok('Game reset.');status();}
  else if(a==='status'||a==='look')status();
  else if(a==='+m')board('M'); else if(a==='+c')board('C');
  else if(a==='-m')unboard('M'); else if(a==='-c')unboard('C');
  else if(a==='load'&&/^\d+$/.test(w[1]||'')&&/^\d+$/.test(w[2]||''))load(+w[1],+w[2]);
  else if((m=line.match(/^(\d+)\s+(\d+)$/)))load(+m[1],+m[2]);
  else er("Unknown command '"+line+"'. Type help.");
}
inp.addEventListener('keydown',e=>{
  if(e.key==='Enter'){const v=inp.value;inp.value='';cmdHist.push(v);hi=cmdHist.length;exec(v);}
  else if(e.key==='ArrowUp'&&hi>0){hi--;inp.value=cmdHist[hi];e.preventDefault();}
  else if(e.key==='ArrowDown'){hi=Math.min(hi+1,cmdHist.length);inp.value=cmdHist[hi]||'';e.preventDefault();}
});
let cmdHist=[],hi=0;
document.querySelectorAll('#mc .keys button').forEach(b=>b.onclick=()=>{exec(b.dataset.c);inp.focus();});
$('mc').addEventListener('click',e=>{if(!window.getSelection().toString())inp.focus();});

/* ---------- 3D scene ---------- */
let THREE_, renderer, scene, camera, water, wbase, boat, bankMats={}, bankMeshes={};
function layout(){
  ['L','R'].forEach(side=>{
    chars.filter(c=>c.loc===side).forEach((c,i)=>{
      const sg=side==='L'?-1:1, col=i%2, row=Math.floor(i/2);
      c.t={x:sg*(4.3+col*1.4),y:0,z:(row-1)*1.7};
    });
  });
  chars.filter(c=>c.loc==='B').forEach((c,i)=>{c.seat=i;});
}
function paintBanks(){
  if(!THREE_)return;
  const lose=fx==='lose', win=fx==='win';
  ['L','R'].forEach(s=>{
    const bad=lose&&unsafe(cnt(s,'M'),cnt(s,'C'));
    bankMats[s].color.setHex(bad?0x8a1f1f:win?0x1f8a45:0x1c5a2e);
  });
}
function mkChar(type){
  const T=THREE_, g=new T.Group(), M=type==='M';
  const col=M?0x6fe3ff:0xff9a1f, mat=new T.MeshStandardMaterial({color:col,flatShading:true,roughness:.7});
  const bg=M?new T.CylinderGeometry(.24,.32,.7,8):new T.ConeGeometry(.4,.8,7);
  const body=new T.Mesh(bg,mat); body.position.y=.38; g.add(body);
  const ln=new T.LineSegments(new T.EdgesGeometry(bg),new T.LineBasicMaterial({color:0xffffff,transparent:true,opacity:.45})); ln.position.y=.38; g.add(ln);
  const head=new T.Mesh(new T.SphereGeometry(.24,10,8),new T.MeshStandardMaterial({color:M?0xf2f7ff:0xff5a3a,flatShading:true})); head.position.y=.98; g.add(head);
  if(M){const h=new T.Mesh(new T.TorusGeometry(.2,.035,6,16),new T.MeshBasicMaterial({color:0xffe14d}));h.rotation.x=Math.PI/2;h.position.y=1.33;g.add(h);}
  else{[-1,1].forEach(s=>{const h=new T.Mesh(new T.ConeGeometry(.07,.28,5),new T.MeshBasicMaterial({color:0xff2d2d}));h.position.set(s*.17,1.27,0);h.rotation.z=-s*.4;g.add(h);});}
  return g;
}
function init(){
  const T=THREE_=window.THREE, view=$('mc-view');
  try{renderer=new T.WebGLRenderer({antialias:true});}catch(e){say('WebGL is unavailable here. You can still play with the terminal.','warn');return;}
  renderer.setPixelRatio(Math.min(devicePixelRatio||1,2)); renderer.setClearColor(0x020a05,1); view.appendChild(renderer.domElement);
  scene=new T.Scene(); scene.fog=new T.Fog(0x020a05,14,34);
  camera=new T.PerspectiveCamera(45,1,.1,100);
  scene.add(new T.AmbientLight(0x4a7a5a,.9)); const dl=new T.DirectionalLight(0xc8ffe0,.9); dl.position.set(4,9,6); scene.add(dl);
  const grid=new T.GridHelper(40,40,0x38ff7a,0x0b4a24); grid.position.y=-1.05; scene.add(grid);
  // banks
  ['L','R'].forEach(s=>{
    const sg=s==='L'?-1:1, geo=new T.BoxGeometry(4.6,1,7.5), m=new T.MeshStandardMaterial({color:0x1c5a2e,flatShading:true});
    const b=new T.Mesh(geo,m); b.position.set(sg*5.3,-.5,0); scene.add(b);
    const e=new T.LineSegments(new T.EdgesGeometry(geo),new T.LineBasicMaterial({color:0x38ff7a})); e.position.copy(b.position); scene.add(e);
    bankMats[s]=m;
  });
  // water
  const wg=new T.PlaneGeometry(6,7.5,20,14); wg.rotateX(-Math.PI/2); wbase=wg.attributes.position.array.slice();
  water=new T.Mesh(wg,new T.MeshStandardMaterial({color:0x0a4a6e,flatShading:true,transparent:true,opacity:.9}));
  water.position.y=-.35; scene.add(water);
  const wl=new T.Mesh(wg,new T.MeshBasicMaterial({color:0x35c9ff,wireframe:true,transparent:true,opacity:.35})); wl.position.y=-.34; water.userData.wl=wl; scene.add(wl);
  // boat
  boat=new T.Group();
  const hg=new T.BoxGeometry(2.3,.45,1.3), hull=new T.Mesh(hg,new T.MeshStandardMaterial({color:0xb9783f,flatShading:true}));
  boat.add(hull); boat.add(new T.LineSegments(new T.EdgesGeometry(hg),new T.LineBasicMaterial({color:0xffd9a0})));
  const bow=new T.Mesh(new T.ConeGeometry(.65,.6,4),new T.MeshStandardMaterial({color:0xb9783f,flatShading:true})); bow.rotation.set(0,Math.PI/4,-Math.PI/2); bow.position.x=1.3; bow.scale.set(1,1,.7); boat.add(bow);
  boat.position.set(1.7,-.15,0); scene.add(boat);
  // characters
  chars.forEach(c=>{c.obj=mkChar(c.type);scene.add(c.obj);});
  layout(); chars.forEach(c=>{c.obj.position.set(c.t.x,0,c.t.z);});
  // orbit
  let th=0, ph=.5, rad=13.5, drag=null;
  view.addEventListener('pointerdown',e=>{drag={x:e.clientX,y:e.clientY};view.style.cursor='grabbing';});
  addEventListener('pointerup',()=>{drag=null;view.style.cursor='grab';});
  addEventListener('pointermove',e=>{if(!drag)return;th=Math.max(-1.2,Math.min(1.2,th-(e.clientX-drag.x)*.006));ph=Math.max(.15,Math.min(1.3,ph+(e.clientY-drag.y)*.005));drag={x:e.clientX,y:e.clientY};});
  function size(){const w=view.clientWidth,h=view.clientHeight;renderer.setSize(w,h,false);camera.aspect=w/h;camera.updateProjectionMatrix();rad=w<560?17:13.5;}
  addEventListener('resize',size); size();
  const clock=new T.Clock(), still=matchMedia('(prefers-reduced-motion:reduce)').matches;
  (function loop(){
    requestAnimationFrame(loop);
    const dt=Math.min(clock.getDelta(),.05), t=clock.elapsedTime, k=1-Math.exp(-4*dt), kc=1-Math.exp(-9*dt);
    // water
    if(!still){const p=water.geometry.attributes.position.array;for(let i=0;i<p.length;i+=3){p[i+1]=Math.sin(wbase[i]*1.1+t*1.6)*.09+Math.cos(wbase[i+2]*1.4+t*1.2)*.08;}
      water.geometry.attributes.position.needsUpdate=true;}
    // boat
    const bx=S.boat==='L'?-1.7:1.7; boat.position.x+=(bx-boat.position.x)*k;
    boat.position.y=-.15+(still?0:Math.sin(t*2)*.05); boat.rotation.z=still?0:Math.sin(t*1.6)*.025;
    boat.scale.x=S.boat==='L'?-1:1;
    // characters
    chars.forEach((c,i)=>{
      let tx,ty,tz;
      if(c.loc==='B'){tx=boat.position.x+(c.seat?.55:-.55);ty=boat.position.y+.22;tz=0;}else{tx=c.t.x;ty=0;tz=c.t.z;}
      const o=c.obj.position, d=Math.hypot(tx-o.x,tz-o.z);
      o.x+=(tx-o.x)*kc; o.z+=(tz-o.z)*kc; o.y+=((ty+Math.min(d,2.5)*.35)-o.y)*kc;
      if(fx==='win'&&!still)o.y=Math.abs(Math.sin(t*5+i))*.5;
      if(fx==='lose'&&!still&&c.loc!=='B'){const bad=unsafe(cnt(c.loc,'M'),cnt(c.loc,'C'));if(bad)o.x=c.t.x+Math.sin(t*40+i)*.06;}
    });
    // camera
    const sway=drag||still?0:Math.sin(t*.25)*.12, a=th+sway;
    camera.position.set(Math.sin(a)*Math.cos(ph)*rad,Math.sin(ph)*rad+.5,Math.cos(a)*Math.cos(ph)*rad);
    camera.lookAt(0,.4,0); renderer.render(scene,camera);
  })();
  paintBanks();
}

/* ---------- boot ---------- */
['M','M','M','C','C','C'].forEach(t=>chars.push({type:t,loc:'R'}));
S={boat:'R',moves:0,over:false}; layout(); hud();
say('MISSIONARIES & CANNIBALS  v1.0','ok');
say('Goal: ferry 3 missionaries (cyan, halo) and 3 cannibals (orange, horns) from RIGHT to LEFT.','ok');
say("Boat holds 1-2 and cannot cross empty. Type 'help' for commands, or try: load 1 1","dim");
status();
const s=document.createElement('script');
s.src='https://cdnjs.cloudflare.com/ajax/libs/three.js/r128/three.min.js';
s.onload=init; s.onerror=()=>wn('Could not load three.js. The terminal still works.');
document.head.appendChild(s);
})();
</script>
"""

display(HTML(GAME_HTML))
````
